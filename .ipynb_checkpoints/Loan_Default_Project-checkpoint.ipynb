# Loan Default Prediction

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    accuracy_score, confusion_matrix, classification_report
)
import joblib
```

```python
# Load the Loan Default dataset
df = pd.read_csv('Loan_default.csv')

df
```

```python
# Basic information
print("First 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nDataset shape:", df.shape)
```

```python
# Check missing values and duplicate rows
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Check the target distribution
print(df["Default"].value_counts())
print("\nTarget percentage:")
print(df["Default"].value_counts(normalize=True).mul(100).round(2))

plt.figure(figsize=(5, 3))
sns.countplot(x=df["Default"])
plt.title("Loan Default Distribution")
plt.xlabel("Default")
plt.ylabel("Count")
plt.show()
```

```python
# Remove the unique LoanID because it does not provide useful predictive information
df = df.drop(columns=["LoanID"])

# Remove duplicate records, if any
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after cleaning:", df.shape)
```

```python
# Numerical columns in the Loan Default dataset
numerical_cols = [
    "Age", "Income", "LoanAmount", "CreditScore",
    "MonthsEmployed", "NumCreditLines", "InterestRate",
    "LoanTerm", "DTIRatio"
]

# Boxplots for numerical features
for col in numerical_cols:
    plt.figure(figsize=(5, 3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
# Clip extreme numerical values using the IQR rule.
# This follows the same outlier-treatment idea as the original project.
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

print(df.head())
```

```python
# Count remaining IQR outliers after clipping
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers")
```

```python
# Separate input features and target
X = df.drop("Default", axis=1)
y = df["Default"]

# Convert categorical columns into numerical dummy variables
categorical_cols = X.select_dtypes(include="object").columns.tolist()
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True, dtype=int)

print("Features after encoding:", X.shape)
print("Encoded columns:")
print(X.columns.tolist())
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# Standardize numerical features for Logistic Regression.
# Tree-based models do not require scaling, but the same scaled data can
# still be used for this project comparison.
sc = StandardScaler()

X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)

print("Scaling completed successfully.")
```

```python
# Logistic Regression
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

ac = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
cm = confusion_matrix(y_test, y_pred)

print("Logistic Regression Results")
print("Accuracy:", ac)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
print("Confusion Matrix:\n", cm)
```

```python
# Train and test scores
train_score = model.score(X_train_scaled, y_train)
test_score = model.score(X_test_scaled, y_test)

print("Train Score:", train_score)
print("Test Score:", test_score)
```

```python
# 5-Fold Cross Validation for Logistic Regression
scores = cross_val_score(
    model,
    X_train_scaled,
    y_train,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("5-Fold CV Scores:", scores)
print("Mean CV Score:", scores.mean())
print("Standard Deviation:", scores.std())
```

```python
# Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

print("Decision Tree Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_dt, zero_division=0))
print("F1-score:", f1_score(y_test, y_pred_dt, zero_division=0))
```

```python
# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_rf, zero_division=0))
print("F1-score:", f1_score(y_test, y_pred_rf, zero_division=0))
```

```python
# AdaBoost
ab_model = AdaBoostClassifier(random_state=42)
ab_model.fit(X_train, y_train)

y_pred_ab = ab_model.predict(X_test)

print("AdaBoost Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_ab))
print("Precision:", precision_score(y_test, y_pred_ab, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_ab, zero_division=0))
print("F1-score:", f1_score(y_test, y_pred_ab, zero_division=0))
```

```python
# Compare all model results in one table
results = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree", "Random Forest", "AdaBoost"],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_ab)
    ],
    "Precision": [
        precision_score(y_test, y_pred, zero_division=0),
        precision_score(y_test, y_pred_dt, zero_division=0),
        precision_score(y_test, y_pred_rf, zero_division=0),
        precision_score(y_test, y_pred_ab, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, y_pred, zero_division=0),
        recall_score(y_test, y_pred_dt, zero_division=0),
        recall_score(y_test, y_pred_rf, zero_division=0),
        recall_score(y_test, y_pred_ab, zero_division=0)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred, zero_division=0),
        f1_score(y_test, y_pred_dt, zero_division=0),
        f1_score(y_test, y_pred_rf, zero_division=0),
        f1_score(y_test, y_pred_ab, zero_division=0)
    ]
})

results.sort_values("F1 Score", ascending=False)
```

```python
# 5-Fold Cross Validation for Random Forest
cv_scores = cross_val_score(
    rf_model,
    X_train,
    y_train,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("5-Fold CV Scores:", cv_scores)
print("Mean CV Accuracy:", cv_scores.mean())
print("Standard Deviation:", cv_scores.std())
```

## Random Forest Hyperparameter Tuning

The grid below is intentionally smaller than a very large search because this dataset contains more than 250,000 rows. It keeps the same tuning concept as the original project while reducing unnecessary computation.

```python
param_grid = {
    "n_estimators": [50, 100],
    "max_depth": [10, 20],
    "min_samples_split": [2, 5]
}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42, n_jobs=-1),
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV Score:", grid_search.best_score_)

best_model = grid_search.best_estimator_
y_pred_best = best_model.predict(X_test)

print("\nTuned Random Forest Test Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_best))
print("Precision:", precision_score(y_test, y_pred_best, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_best, zero_division=0))
print("F1-score:", f1_score(y_test, y_pred_best, zero_division=0))
```

```python
# Confusion Matrix for the tuned Random Forest
cm = confusion_matrix(y_test, y_pred_best)

plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False)
plt.title("Confusion Matrix - Tuned Random Forest")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()
```

```python
# Classification report
print(classification_report(
    y_test,
    y_pred_best,
    target_names=["No Default", "Default"],
    zero_division=0
))
```

```python
# Feature importance from the tuned Random Forest
feature_importance = pd.Series(
    best_model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

plt.figure(figsize=(10, 6))
feature_importance.head(15).sort_values().plot(kind="barh")
plt.title("Top 15 Feature Importances - Tuned Random Forest")
plt.xlabel("Importance")
plt.show()

print(feature_importance.head(15))
```

```python
# Save the trained model, scaler, and feature names.
# The feature names are required when making predictions on new data.

joblib.dump(best_model, "loan_default_model.pkl")
joblib.dump(sc, "loan_default_scaler.pkl")
joblib.dump(X_train.columns.tolist(), "loan_default_features.pkl")

print("Model, scaler, and feature list saved successfully!")
```
